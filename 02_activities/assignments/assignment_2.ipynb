#### Starter Code for Question 3


```python
#from typing import List

def missing_num(nums: List) -> int:
    # TODO
    missings = []

    for x in range(0, int(max(nums))):
        if x not in nums:
            missings.append(x)
    if missings:
        return missings
    else:
        return -1
  
```

```python
# Your answer here
Input: s = "{([][])}"
Output: True
Explanation: every open bracket is closed by the same type of the bracket and open brackets are closed in correct order.


Input: s = "([{])"
Output: False
Explanation: There is an open bracket which is not correctly closed.
```


-   Copy the solution your partner wrote. 


```python
# Your answer here

def is_valid_brackets(s: str) -> bool:
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}

    for char in s:
        if char in pairs.values():  # opening bracket
            stack.append(char)
        elif char in pairs:  # closing bracket
            if not stack or stack[-1] != pairs[char]:
                return False
            stack.pop()
        else:
            return False  # invalid character

    return len(stack) == 0
```


-   Explain why their solution works in your own words.


```python
# Your answer here
The function is_valid_brackets checks every character in a given string "s" and return true if the string has valid sequence of brackets.
First, an empty list 'stack' is defined to track the opening bracket in "s".
Then, the dictionary "pairs" is defined where every "key" is a clossing bracket and every corresponding "value" is its correct closing bracket.
Next the code checks every character in "s"
If the code finds an oppening bracket equivalent to the value of the dictionary "pairs", it adds that bracket into the list "stack", until "stack" contains all the openning brackets from the string"s".
Then the code checks the keys in "pairs".
Then it verifies if the stack is empy (conditiion 1) or the last element in stack (i.e last openning bracket) does not match with the current checked key (closing brackets) in'pairs (condition 2).
If either one the two conditions is satisfied, the code return False, meaning that the sequence of brackets is invalid.
If none of the two condition is satisfied, a valid pair is found, and the last element in stack is remove (poped) from stack.This follows Stack principle of Last-In First-Out. 
If the code finds a character which is not in pairs, it return invalid sequence of brackets.
Finally, when the stack gets empty, it means that all openning brackets has found their macthing clossing brackets. Thus, the string has a valid sequence of brackets.


 
```
